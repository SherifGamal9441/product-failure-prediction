```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# 1. Setup Data
X = df_train.drop(['failure', 'product_code'], axis=1)
y = df_train['failure']
groups = df_train['product_code'] # Vital for GroupKFold

# 2. Identify Column Types
numeric_features = [c for c in X.columns if X[c].dtype in ['float64', 'int64']]
categorical_features = ['attribute_0', 'attribute_1'] # attribute_0/1 are objects

# 3. Create Preprocessing Pipeline
# For numeric: Impute missing values (mean for now) -> Scale
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')), 
    ('scaler', StandardScaler())
])

# For categorical: One-hot encode
categorical_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# 4. The Model: Logistic Regression often beats Trees on this specific type of data
model = Pipeline(steps=[('preprocessor', preprocessor),
                        ('classifier', LogisticRegression(C=0.01, solver='liblinear'))])

# 5. Validation Strategy: GroupKFold
# We split so that products in Validation are NOT in Train
cv = StratifiedGroupKFold(n_splits=5)

scores = []
for fold, (train_idx, val_idx) in enumerate(cv.split(X, y, groups)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model.fit(X_train, y_train)
    
    val_preds = model.predict_proba(X_val)[:, 1]
    score = roc_auc_score(y_val, val_preds)
    scores.append(score)
    print(f"Fold {fold+1} AUC: {score:.4f}")

print(f"\nAverage AUC: {np.mean(scores):.4f}")
```

```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer  # Explicitly enable
from sklearn.impute import IterativeImputer
from sklearn.preprocessing import PowerTransformer, OneHotEncoder, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import roc_auc_score

# 1. Setup Data
X = df_train.drop(['failure', 'product_code'], axis=1)
y = df_train['failure']
groups = df_train['product_code']

# 2. Define Features
# We want to log-transform 'loading' specifically
X['loading'] = np.log1p(X['loading']) 

numeric_features = [c for c in X.columns if X[c].dtype in ['float64', 'int64']]
categorical_features = ['attribute_0', 'attribute_1']

# 3. Advanced Preprocessing
# PowerTransformer handles skewness better than StandardScaler
# IterativeImputer uses correlations to fill missing values
numeric_transformer = Pipeline(steps=[
    ('imputer', IterativeImputer(max_iter=10, random_state=0)), 
    ('scaler', PowerTransformer()) 
])

categorical_transformer = Pipeline(steps=[
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# 4. Model: Same Logistic Regression, but the data feeding it is much cleaner
# Increased C slightly to reduce regularization (allow model to learn more complex patterns)
model = Pipeline(steps=[('preprocessor', preprocessor),
                        ('classifier', LogisticRegression(C=0.05, solver='liblinear', penalty='l1'))])

# 5. Validation
cv = StratifiedGroupKFold(n_splits=5)
scores = []

print("--- Running Improved Linear Pipeline ---")
for fold, (train_idx, val_idx) in enumerate(cv.split(X, y, groups)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model.fit(X_train, y_train)
    
    val_preds = model.predict_proba(X_val)[:, 1]
    score = roc_auc_score(y_val, val_preds)
    scores.append(score)
    print(f"Fold {fold+1} AUC: {score:.4f}")

print(f"\nAverage AUC: {np.mean(scores):.4f}")
```

```python
import numpy as np
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# ---------------------------------------------------------
# 1. SETUP DATA (Assuming X_train / y_train exist)
# ---------------------------------------------------------
# We combine them back temporarily to apply feature engineering uniformly
# (If you have a separate test set, apply the same logic there!)
X = X_train.copy()
y = y_train.copy()

# ---------------------------------------------------------
# 2. FEATURE ENGINEERING: ADD TOP INTERACTIONS
# ---------------------------------------------------------
# Based on your SHAP list, we create the unique pairs.
# measurement_13 is the "Hub" feature.

def add_interactions(df):
    df_eng = df.copy()
    
    # The "Hub" Interactions (measurement_13 vs others)
    df_eng['inter_13_10'] = df_eng['measurement_13'] * df_eng['measurement_10']
    df_eng['inter_13_load'] = df_eng['measurement_13'] * df_eng['loading']
    df_eng['inter_13_17'] = df_eng['measurement_13'] * df_eng['measurement_17']
    df_eng['inter_13_7']  = df_eng['measurement_13'] * df_eng['measurement_7']
    df_eng['inter_13_14'] = df_eng['measurement_13'] * df_eng['measurement_14']
    df_eng['inter_13_4']  = df_eng['measurement_13'] * df_eng['measurement_4']
    df_eng['inter_13_5']  = df_eng['measurement_13'] * df_eng['measurement_5']
    df_eng['inter_13_9']  = df_eng['measurement_13'] * df_eng['measurement_9']
    df_eng['inter_13_2']  = df_eng['measurement_13'] * df_eng['measurement_2']
    
    # Other Interactions found in the list
    df_eng['inter_7_load'] = df_eng['measurement_7'] * df_eng['loading']
    
    return df_eng

print("Generating interaction features...")
X_eng = add_interactions(X)

# ---------------------------------------------------------
# 3. SELECT FEATURES (Base + New Interactions)
# ---------------------------------------------------------
# We keep the "Main Effect" features + our new Interaction columns
base_features = [
    'loading', 'measurement_13', 'measurement_17', 'measurement_5',
    'measurement_14', 'measurement_9', 'measurement_10', 'measurement_4',
    'measurement_7', 'measurement_2', 'measurement_0'
]

interaction_features = [
    'inter_13_10', 'inter_13_load', 'inter_13_17', 'inter_13_7',
    'inter_13_14', 'inter_13_4', 'inter_13_5', 'inter_13_9',
    'inter_13_2', 'inter_7_load'
]

final_features = base_features + interaction_features
X_final = X_eng[final_features]

print(f"Final feature count: {len(final_features)}")

# ---------------------------------------------------------
# 4. STRATIFIED K-FOLD CV LOOP
# ---------------------------------------------------------
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_preds = np.zeros(len(X_final))
fold_scores = []

print("\nStarting 5-Fold Cross-Validation...")

for fold, (train_idx, val_idx) in enumerate(kf.split(X_final, y)):
    # Split data
    X_tr, X_val_fold = X_final.iloc[train_idx], X_final.iloc[val_idx]
    y_tr, y_val_fold = y.iloc[train_idx], y.iloc[val_idx]
    
    # Preprocessing (Impute/Scale inside the fold to prevent leakage)
    # Note: XGBoost handles NaNs, but scaling interactions can sometimes help convergence
    # We will use raw values here as XGBoost is robust.
    
    # Define Model (XGBoost 2.0.3 compatible)
    clf = xgb.XGBClassifier(
        n_estimators=500,        # Increased slightly for more features
        learning_rate=0.05,      # Lower learning rate for stability
        max_depth=4,             # Keep trees shallow to prevent overfitting interactions
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1,
        eval_metric='auc',
        early_stopping_rounds=50
    )
    
    # Train
    clf.fit(
        X_tr, y_tr, 
        eval_set=[(X_val_fold, y_val_fold)], 
        verbose=False
    )
    
    # Predict
    val_preds = clf.predict_proba(X_val_fold)[:, 1]
    oof_preds[val_idx] = val_preds
    
    # Score
    score = roc_auc_score(y_val_fold, val_preds)
    fold_scores.append(score)
    print(f"Fold {fold+1} AUC: {score:.4f}")

# ---------------------------------------------------------
# 5. RESULTS
# ---------------------------------------------------------
print("\n--- Final Results ---")
print(f"Average AUC: {np.mean(fold_scores):.4f}")
print(f"Standard Dev: {np.std(fold_scores):.4f}")
```

```python
from tensorflow.keras import layers, models, callbacks, optimizers
import tensorflow as tf

# --- CONFIGURATION DICTIONARY ---
architectures = {
    "5.1 Modern Heavy (Swish)": {
        "type": "swish_heavy"
    },
    "5.2 Wide & Deep": {
        "type": "wide_deep"
    },
    "5.3 ResNet-512": {
        "type": "resnet"
    }
}

def build_modern_heavy(input_shape):
    # Same as 3.2, but with Swish
    inputs = layers.Input(shape=input_shape)
    x = inputs
    for _ in range(5):
        x = layers.Dense(512, activation='swish')(x)
        x = layers.Dropout(0.6)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)
    return models.Model(inputs, outputs)

def build_wide_and_deep(input_shape):
    inputs = layers.Input(shape=input_shape)
    
    # Deep Path (The Heavy Net)
    x = inputs
    for _ in range(5):
        x = layers.Dense(512, activation='relu')(x)
        x = layers.Dropout(0.6)(x)
        
    # Wide Path: Concatenate Deep Output with Original Input
    # This gives the final layer access to raw features + deep features
    combined = layers.concatenate([x, inputs])
    
    outputs = layers.Dense(1, activation='sigmoid')(combined)
    return models.Model(inputs, outputs)

def build_resnet(input_shape):
    inputs = layers.Input(shape=input_shape)
    
    # Project input to 512 dims so we can add it to dense layers
    x = layers.Dense(512, activation='relu')(inputs)
    x = layers.Dropout(0.5)(x)
    
    # 4 Residual Blocks
    for _ in range(4):
        # Save input (Identity)
        shortcut = x
        
        # Dense Layer
        x = layers.Dense(512, activation='relu')(x)
        x = layers.Dropout(0.5)(x)
        
        # ADD connection (The Residual Magic)
        x = layers.Add()([x, shortcut])
        
    outputs = layers.Dense(1, activation='sigmoid')(x)
    return models.Model(inputs, outputs)

# --- EXPERIMENT LOOP ---
results = {}

print(f"{'Architecture':<30} | {'Mean AUC':<10} | {'Std Dev':<10}")
print("-" * 55)

for name, config in architectures.items():
    cv_scores = []
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    
    for train_idx, val_idx in skf.split(X_scaled, y_target):
        X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
        y_train, y_val = y_target[train_idx], y_target[val_idx]
        
        # Select builder based on config type
        if config['type'] == 'swish_heavy':
            model = build_modern_heavy((input_dim,))
        elif config['type'] == 'wide_deep':
            model = build_wide_and_deep((input_dim,))
        elif config['type'] == 'resnet':
            model = build_resnet((input_dim,))
            
        model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['AUC'])
        
        # Early Stopping
        es = callbacks.EarlyStopping(
            monitor='val_AUC', 
            mode='max', 
            verbose=0, 
            patience=15, 
            restore_best_weights=True
        )
        
        model.fit(X_train, y_train, 
                  epochs=100, 
                  batch_size=128, 
                  validation_data=(X_val, y_val), 
                  callbacks=[es],
                  verbose=0)
        
        preds = model.predict(X_val, verbose=0).flatten()
        score = roc_auc_score(y_val, preds)
        cv_scores.append(score)
        
    mean_auc = np.mean(cv_scores)
    std_auc = np.std(cv_scores)
    
    results[name] = (mean_auc, std_auc)
    print(f"{name:<30} | {mean_auc:.4f}     | {std_auc:.4f}")
```

```python
# --- CONFIGURATION DICTIONARY ---
architectures = {
    "6.1 The Transformer (GELU)": {
        "type": "gelu"
    },
    "6.2 SNN (SELU + AlphaDrop)": {
        "type": "snn"
    },
    "6.3 Tuned Swish (Scheduler)": {
        "type": "swish_scheduler"
    }
}

def build_gelu_model(input_shape):
    # Rival to Swish
    model = models.Sequential()
    model.add(layers.InputLayer(input_shape=input_shape))
    for _ in range(5):
        model.add(layers.Dense(512, activation='gelu'))
        model.add(layers.Dropout(0.6))
    model.add(layers.Dense(1, activation='sigmoid'))
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['AUC'])
    return model, [] # No extra callbacks

def build_snn_model(input_shape):
    # Self-Normalizing Network (The Theoretical King of Tabular Data)
    model = models.Sequential()
    model.add(layers.InputLayer(input_shape=input_shape))
    for _ in range(5):
        # SNN REQUIREMENT 1: Activation must be 'selu'
        # SNN REQUIREMENT 2: Kernel Initializer must be 'lecun_normal'
        model.add(layers.Dense(512, activation='selu', kernel_initializer='lecun_normal'))
        
        # SNN REQUIREMENT 3: Must use AlphaDropout, not regular Dropout
        model.add(layers.AlphaDropout(0.1)) # AlphaDropout requires lower rates (0.05-0.1)
        
    model.add(layers.Dense(1, activation='sigmoid'))
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['AUC'])
    return model, []

def build_swish_scheduler_model(input_shape):
    # Same as 5.1 (The Winner)
    model = models.Sequential()
    model.add(layers.InputLayer(input_shape=input_shape))
    for _ in range(5):
        model.add(layers.Dense(512, activation='swish'))
        model.add(layers.Dropout(0.6))
    model.add(layers.Dense(1, activation='sigmoid'))
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['AUC'])
    
    # Extra Callback: Reduce LR when stuck
    lr_scheduler = callbacks.ReduceLROnPlateau(
        monitor='val_AUC', 
        factor=0.5,      # Cut LR in half
        patience=5,      # If no improvement for 5 epochs
        min_lr=1e-6, 
        verbose=0,
        mode='max'
    )
    return model, [lr_scheduler]

# --- EXPERIMENT LOOP ---
results = {}

print(f"{'Architecture':<30} | {'Mean AUC':<10} | {'Std Dev':<10}")
print("-" * 55)

for name, config in architectures.items():
    cv_scores = []
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    
    for train_idx, val_idx in skf.split(X_scaled, y_target):
        X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
        y_train, y_val = y_target[train_idx], y_target[val_idx]
        
        # 1. Build Model & Get Config-Specific Callbacks
        if config['type'] == 'gelu':
            model, extra_callbacks = build_gelu_model((input_dim,))
        elif config['type'] == 'snn':
            model, extra_callbacks = build_snn_model((input_dim,))
        elif config['type'] == 'swish_scheduler':
            model, extra_callbacks = build_swish_scheduler_model((input_dim,))
            
        # 2. Setup Callbacks (Standard ES + Any Extra)
        es = callbacks.EarlyStopping(
            monitor='val_AUC', 
            mode='max', 
            verbose=0, 
            patience=20, 
            restore_best_weights=True
        )
        my_callbacks = [es] + extra_callbacks
        
        # 3. Train
        model.fit(X_train, y_train, 
                  epochs=100, 
                  batch_size=128, 
                  validation_data=(X_val, y_val), 
                  callbacks=my_callbacks,
                  verbose=0)
        
        # 4. Score
        preds = model.predict(X_val, verbose=0).flatten()
        score = roc_auc_score(y_val, preds)
        cv_scores.append(score)
        
    mean_auc = np.mean(cv_scores)
    std_auc = np.std(cv_scores)
    
    results[name] = (mean_auc, std_auc)
    print(f"{name:<30} | {mean_auc:.4f}     | {std_auc:.4f}")
```

```python
import optuna
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score

# --- OPTUNA OBJECTIVE FUNCTION ---
def objective(trial):
    # 1. Suggest Parameters
    # Log scale is important for C because 0.1 and 1.0 is a bigger difference than 10.1 and 11.0
    c_param = trial.suggest_float("C", 1e-3, 100.0, log=True)
    kernel = trial.suggest_categorical("kernel", ["rbf", "sigmoid"])
    gamma = trial.suggest_categorical("gamma", ["scale", "auto"])
    
    # 2. Cross Validation (3-Fold for Speed during search)
    skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    cv_scores = []
    
    for train_idx, val_idx in skf.split(X_scaled, y_target):
        X_train, X_val = X_scaled[train_idx], X_scaled[val_idx]
        y_train, y_val = y_target[train_idx], y_target[val_idx]
        
        # Train SVM (Fast mode: probability=False)
        # We use decision_function() for AUC which is much faster
        clf = SVC(
            C=c_param, 
            kernel=kernel, 
            gamma=gamma, 
            probability=False, 
            random_state=42,
            cache_size=1000 # Use 1GB RAM for cache to speed it up
        )
        
        clf.fit(X_train, y_train)
        
        # Predict
        preds = clf.decision_function(X_val)
        score = roc_auc_score(y_val, preds)
        cv_scores.append(score)
        
    return np.mean(cv_scores)

# --- RUN OPTIMIZATION ---
# Turn off Optuna's massive logging block, just show progress bar
optuna.logging.set_verbosity(optuna.logging.WARNING)

print("Starting Optuna Search (20 Trials)...")
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=20, show_progress_bar=True)

print("-" * 50)
print(f"Best CV Score: {study.best_value:.5f}")
print("Best Params:   ", study.best_params)
```
